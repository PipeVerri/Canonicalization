```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm
from ipywidgets import interact, IntSlider

x = np.linspace(-5, 5, 1000)

def plot_t(v):
    plt.figure(figsize=(8, 5))
    
    # Distribución t
    plt.plot(x, t.pdf(x, df=v), label=f"t(df={v})", linewidth=2)

    # Normal estándar como referencia
    plt.plot(x, norm.pdf(x), "--", label="Normal(0,1)", alpha=0.7)

    plt.title("Distribución t de Student")
    plt.xlabel("x")
    plt.ylabel("Densidad")
    plt.xlim(-5, 5)
    plt.ylim(0, 0.45)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

interact(
    plot_t,
    v=IntSlider(
        value=5,
        min=1,
        max=100,
        step=1,
        description="ν"
    )
);
```
