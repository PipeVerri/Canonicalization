```python
from pathlib import Path
from tqdm.auto import tqdm

genomic_benchmark_dir = Path("../../data/genomic_benchmark")

max_lengths = {dataset.name: 0 for dataset in genomic_benchmark_dir.iterdir() if dataset.is_dir()}
train_lengths = {k: 0 for k in max_lengths.keys()}

for dataset in max_lengths.keys():
    files = list((genomic_benchmark_dir / dataset).glob("**/*.txt"))
    train_lengths[dataset] = len(files)
    for file in tqdm(files, desc=f"Processing {dataset}", total=len(files)):
        with open(file, "r") as f:
            max_lengths[dataset] = max(max_lengths[dataset], len(f.read().strip()))
```
