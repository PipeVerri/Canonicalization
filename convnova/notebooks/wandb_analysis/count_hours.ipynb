```python
import wandb
api = wandb.Api()
runs = api.runs("pipe-verri-gh/Canonicalization", filters={
    "group": "baseline"
})

# Get the total runtime for the group
total_runtime = sum(run.summary.get("_wandb", {}).get("runtime", 0) for run in runs)
print(f"Total runtime for group 'baseline': {total_runtime} seconds")
```

```python
from datetime import timedelta
duration = timedelta(seconds=total_runtime)
print(f"Total runtime for group 'baseline': {duration}")
```
