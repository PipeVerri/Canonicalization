```python
from Bio import SeqIO
import json

g1 = "../data/hg38/hg38.ml.fa"
g1_records = {}
for record in SeqIO.parse(g1, "fasta"):
    g1_records[record.id] = record.seq 
```

```python
import re

g2 = "../data/hg38/GRCh38.primary_assembly.genome.fa"
for record in SeqIO.parse(g2, "fasta"):
    if record.id in g1_records:
        g1_seq = str(g1_records[record.id]).lower()
        g2_seq = str(record.seq).lower()
        g2_seq = re.sub(r"[^atcgn]", "n", g2_seq) # Replace everyything that's not ACGTN with N
        print(g1_seq == g2_seq)
```
